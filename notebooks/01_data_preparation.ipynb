```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/final_labeled_reports.csv")
print(f"Loaded {len(df)} reports")
print(f"Columns: {df.columns.tolist()}")
```

```python
from sklearn.model_selection import train_test_split

# Standard split: 70% train, 10% validation, 20% test
# Follows Chen et al. (2020) for OpenI
# random_state=42 ensures reproducibility

train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.6667, random_state=42)

print(f"Train: {len(train_df)} reports ({len(train_df)/len(df)*100:.1f}%)")
print(f"Val: {len(val_df)} reports ({len(val_df)/len(df)*100:.1f}%)")
print(f"Test: {len(test_df)} reports ({len(test_df)/len(df)*100:.1f}%)")
```

```python
train_df.to_csv("train_set.csv", index=False)
val_df.to_csv("val_set.csv", index=False)
test_df.to_csv("test_set.csv", index=False)

print("Saved all three sets")

from google.colab import files
files.download("train_set.csv")
files.download("val_set.csv")
files.download("test_set.csv")
```

```python
# Check that no report appears in more than one set
train_ids = set(train_df['id'])
val_ids = set(val_df['id'])
test_ids = set(test_df['id'])

print(f"Train ∩ Val: {len(train_ids & val_ids)} (should be 0)")
print(f"Train ∩ Test: {len(train_ids & test_ids)} (should be 0)")
print(f"Val ∩ Test: {len(val_ids & test_ids)} (should be 0)")
```

```python
#check label distribution
pathology_columns = [
    'No Finding', 'Enlarged Cardiomediastinum', 'Cardiomegaly',
    'Lung Lesion', 'Lung Opacity', 'Edema', 'Consolidation',
    'Pneumonia', 'Atelectasis', 'Pneumothorax', 'Pleural Effusion',
    'Pleural Other', 'Fracture', 'Support Devices'
]

print("Label distribution across splits:")
for col in pathology_columns:
    train_pos = (train_df[col] == 1.0).sum()
    val_pos = (val_df[col] == 1.0).sum()
    test_pos = (test_df[col] == 1.0).sum()
    print(f"{col:30s} | Train: {train_pos:>4} | Val: {val_pos:>3} | Test: {test_pos:>4}")
```
